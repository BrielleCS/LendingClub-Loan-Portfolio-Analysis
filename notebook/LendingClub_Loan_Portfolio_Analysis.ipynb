# LendingClub Loan Portfolio Analysis: CECL Credit Risk Modeling, Returns, and Portfolio Concentration

## Overview
This analysis examines over 2.2M loans from the LendingClub accepted loans dataset issued between 2007–2018 to evaluate credit risk, loan performance, and portfolio concentration. To measure realized portfolio outcomes, the dataset was filtered to approximately 1.3M loans with finalized statuses — specifically Fully Paid and Charged Off loans — excluding loans still in repayment or other non-finalized statuses.

Using SQL for aggregation, Python for preprocessing and modeling, and Tableau for dashboard development, the analysis focuses on key business questions related to borrower risk, portfolio returns, and concentration exposure.

The analysis identifies high-risk borrower segments, evaluates risk-adjusted returns across credit grades, and examines how loan exposure is distributed across borrower categories and geographic regions. 

In addition, a credit risk modeling layer is introduced using a Current Expected Credit Loss (CECL) framework, incorporating Probability of Default (PD), Loss Given Default (LGD), and Exposure at Default (EAD) to quantify expected credit losses (EL) across borrower segments. This enhancement enables a forward-looking risk view of the portfolio, complementing realized performance metrics with modeled credit risk exposure.

The workflow demonstrates end-to-end analytical and visualization skills across the modern data analytics stack.

## Data Preparation

This section inspects the LendingClub dataset to understand its structure, size, and key variables before proceeding with this analysis.

Due to the large dataset size (2.2M+ rows), preprocessing was performed using a memory-efficient chunking approach in Python. The filtered dataset was then cleaned and prepared for all subsequent analyses.

### Key Steps:
- Loaded the LendingClub accepted loans dataset (151 columns) covering 2007–2018
- Applied filtering and column selection during chunked processing prior to exporting the reduced dataset
    - Selected 22 relevant columns for analyzing credit risk, returns, and portfolio concentration
    - Processed the dataset in 100k-row chunks to efficiently manage memory usage during preprocessing
    - Filtered the dataset to include only `Fully Paid` and `Charged Off` loans to ensure accurate measurement of default rates and realized returns, resulting in approximately 1.3M records  
- Handled missing values by imputing the categorical variable (`emp_length`) with `Unknown`, filling missing numeric values (`recoveries` and `collection_recovery_fee`) with `0`, and retaining missing dates (`last_pymnt_d`) as `NaT`, where appropriate for time-based analysis
- Converted date fields to datetime format and verified consistency across key numeric and categorical variables
- Performed validation checks on row counts and loan status distributions to ensure data consistency

```python
# -------------------------------
# Memory-Efficient CSV Filter
# -------------------------------

# Input and output paths
input_file = 'accepted_2007_to_2018Q4.csv'   # original large CSV
output_file = 'accepted_loans_filtered.csv'     # new filtered CSV

# Columns to keep
columns_to_keep = [
    'id', 'loan_amnt', 'term', 'int_rate', 'installment',
    'grade', 'sub_grade', 'emp_length', 'home_ownership', 
    'annual_inc', 'issue_d', 'loan_status', 'purpose', 
    'addr_state', 'fico_range_low', 'fico_range_high', 
    'total_pymnt', 'total_rec_prncp', 'total_rec_int',  
    'last_pymnt_d', 'recoveries', 'collection_recovery_fee'
]

# Chunk size (100k rows at a time)
chunksize = 100_000

# Process chunks and write to new CSV
first_chunk = True  # to write header only once

for chunk in pd.read_csv(input_file, chunksize=chunksize, usecols=columns_to_keep):
    
    # filter for only 'Fully Paid' and 'Charged Off' loans
    chunk = chunk[chunk['loan_status'].isin(['Fully Paid', 'Charged Off'])]    
    
    # Append to new CSV
    chunk.to_csv(output_file, mode='w' if first_chunk else 'a', index=False, header=first_chunk)

    # after first chunk, header is no longer written
    first_chunk = False  

print(f"Filtered CSV saved as '{output_file}' successfully!")
```

## Dashboard 2: Default Contribution & Risk Distribution

### Default Contribution Dataset (Pareto Construction)
Default contribution of realized losses across grades and subgrades used for Pareto analysis to assess whether a small number of grades and subgrades account for most of the total losses.

```python
query = """
WITH defaults AS (
     SELECT 
        grade, sub_grade,
        SUM(loan_amnt)::numeric AS default_amount
     FROM accepted_loans
     WHERE loan_status = 'Charged Off'
     GROUP BY grade, sub_grade
     ),
   cum_sum AS (
    SELECT
        grade, sub_grade,
        default_amount,
        default_amount * 1.0 / SUM(default_amount) OVER () * 100 AS default_contribution_pct,
        SUM(default_amount) OVER (ORDER BY default_amount DESC) / SUM(default_amount) OVER () * 100 AS cumulative_pct
    FROM defaults
    )
SELECT
    grade, sub_grade,
    ROUND(default_amount, 2) AS default_amount,
    ROUND(default_contribution_pct, 2) AS default_contribution_pct,
    ROUND(cumulative_pct, 2) AS cumulative_pct
FROM cum_sum
ORDER BY default_amount DESC
"""

# SQL execution (kept for transparency; not executed in final notebook)
# df_default_contribution = pd.read_sql(query, engine)
# df_default_contribution.to_csv("../data/default_contribution_pareto.csv", index=False)

df_default_contribution = pd.read_csv("../data/default_contribution_pareto.csv")
```

### Portfolio Concentration & Inequality (Lorenz Curve & Gini Coefficient)

Loan exposure distribution across borrowers is visualized using a Lorenz curve, highlighting the degree of concentration within the portfolio for Fully Paid and Charged Off loans. The Gini coefficient is computed to quantify inequality in loan exposure distribution, providing a single metric of concentration.

  The Gini Coefficient ranges from:\
    0 → Perfect equality (loan exposure evenly distributed across all borrowers)\
    1 → Perfect concentration (all loan exposure held by a single borrower)


```python
# Load cleaned dataset
df_lorenz = pd.read_csv("../data/accepted_loans_clean.csv")

# Prepare Lorenz curve data
loan_amnt_sorted = np.sort(df_lorenz['loan_amnt'])
cumulative_amnt = np.cumsum(loan_amnt_sorted)
cumulative_amnt = cumulative_amnt / cumulative_amnt[-1]
cumulative_id = np.arange(1, len(loan_amnt_sorted)+1) / len(loan_amnt_sorted)

# Plot Lorenze Curve
plt.figure(figsize=(12,6))
plt.plot(cumulative_id, cumulative_amnt, label='Lorenz Curve', color='blue')
plt.plot([0,1], [0,1], linestyle='--', color='gray', label='Perfect Equality')

# Shading Inequality Area
plt.fill_between(cumulative_id, cumulative_amnt, cumulative_id, color='orange', alpha=0.3, label='Inequality Area')

plt.title('Lorenz Curve - Loan Portfolio Concentration', fontsize=14, fontweight='bold', pad=15)
plt.xlabel('Cumulative Share of Borrowers', fontsize=12, fontweight='bold', labelpad=15)
plt.ylabel('Cumulative Share of Loan Amount', fontsize=12, fontweight='bold', labelpad=15)
plt.legend()
plt.grid(True, linestyle='--', alpha=0.5)

# Compute and plot Gini coefficient
gini = 1 - 2 * np.trapz(cumulative_amnt, cumulative_id)
plt.text(0.6, 0.2, f'Gini Coefficient: {gini:.3f}', fontsize=11, bbox=dict(facecolor='white', alpha=0.6))

#plt.savefig("../images/lorenz_curve.png", dpi=300, bbox_inches="tight")
#plt.show()

# Construct and save Lorenz curve dataset for later Tableau visualization
lorenz_df = pd.DataFrame({"cumulative_borrowers": cumulative_id, "cumulative_loan_amount": cumulative_amnt})
lorenz_df.to_csv("../data/lorenz_curve.csv", index=False)

# Save Gini coefficient
gini_df = pd.DataFrame({"gini_coefficient": [gini]})
gini_df.to_csv("../data/gini_coefficient.csv", index=False)
```

### Credit Risk Modeling & Expected Loss (CECL Framework)
Loan-level data is processed to construct segment-based credit risk measures. The workflow involves:

- Identifying defaulted loans using loan status flags
- Aggregating borrower segments by grade for risk segmentation
- Computing exposure and loss measures at the loan level
- Rolling up results to the grade level for portfolio comparison
- Combining risk components into a unified expected loss framework

The final output is a segmented credit risk table used to assess loss distribution across borrower grades.

```python
# ============================================================
# 1. PROBABILITY OF DEFAULT (PD)
# ============================================================
# Create default indicator:
# Charged Off = 1
# Fully Paid = 0

df['is_default'] = (df['loan_status'] == 'Charged Off').astype(int)

# Calculate historical PD by grade
pd_table = (df.groupby('grade').agg(total_loans=('loan_status', 'count'), 
                                    default_loans=('is_default', 'sum'),
                                    PD=('is_default', 'mean')).reset_index())

# ============================================================
# 2. CREATE DEFAULT-ONLY DATASET
# ============================================================
# LGD and EAD should only be calculated on defaulted loans

df_defaults = df[df['loan_status'] == 'Charged Off'].copy()

# ============================================================
# 3. EXPOSURE AT DEFAULT (EAD)
# ============================================================
# Approximation: Original Loan Amount - Principal Repaid

df_defaults['EAD'] = (df_defaults['loan_amnt'] - df_defaults['total_rec_prncp'])

# Prevent negative values
df_defaults['EAD'] = df_defaults['EAD'].clip(lower=0)

# ============================================================
# 4. LOSS GIVEN DEFAULT (LGD)
# ============================================================
# Realized Loss = Loan Amount - Principal Recovered - Recoveries

df_defaults['loss'] = (df_defaults['loan_amnt'] - df_defaults['total_rec_prncp'] - df_defaults['recoveries'])

# Prevent negative losses
df_defaults['loss'] = df_defaults['loss'].clip(lower=0)

# LGD = Loss / EAD

df_defaults['LGD'] = (df_defaults['loss'] / df_defaults['EAD'])

# Handle divide-by-zero cases
df_defaults['LGD'] = (df_defaults['LGD'].replace([np.inf, -np.inf], np.nan).fillna(0))

# ============================================================
# 5. AGGREGATE EAD & LGD BY GRADE
# ============================================================

risk_table = (df_defaults.groupby('grade').agg(EAD=('EAD', 'mean'), LGD=('LGD', 'mean')).reset_index())

# ============================================================
# 6. MERGE PD + EAD + LGD
# ============================================================

cecl_table = pd.merge(pd_table, risk_table, on='grade', how='left')

# ============================================================
# 7. EXPECTED LOSS (EL)
# ============================================================
# EL = PD × LGD × EAD

cecl_table['EL'] = (cecl_table['PD'] * cecl_table['EAD'] * cecl_table['LGD'])

# ============================================================
# 8. FINAL OUTPUT TABLE
# ============================================================

cecl_table = cecl_table[['grade', 'total_loans', 'default_loans', 'PD', 'EAD', 'LGD', 'EL']]

# Sort highest expected loss first
cecl_table = cecl_table.sort_values(by='EL', ascending=False)

# Save CECL Table
cecl_table.to_csv("../data/cecl_table.csv", index=False)
```

```python
# ============================================================
# FORMAT CECL TABLE AND DISPLAY
# ============================================================
cecl_display = cecl_table.copy()

# Convert PD and LGD to percentages
cecl_display['PD'] = (cecl_display['PD'] * 100).round(2).astype(str) + '%'
cecl_display['LGD'] = (cecl_display['LGD'] * 100).round(2).astype(str) + '%'

# Round dollar-based values
cecl_display['EAD'] = cecl_display['EAD'].round(0).astype(int)
cecl_display['EAD'] = cecl_display['EAD'].apply(lambda x: f"${x:,.0f}")

cecl_display['EL'] = cecl_display['EL'].round(0).astype(int)
cecl_display['EL'] = cecl_display['EL'].apply(lambda x: f"${x:,.0f}")

# Display formatted table
cecl_display.style.set_table_styles([{'selector': 'th', 'props': [('padding', '10px 15px'), ('text-align', 'center')]},
                                     {'selector': 'td', 'props': [('padding', '10px 15px'), ('text-align', 'center')]}])

fig, ax = plt.subplots(figsize=(12, 4))
ax.axis('off')

ax.set_title("CECL Credit Risk Summary Table by Grade", fontsize=16, fontweight='bold', pad=5)

table = ax.table(cellText=cecl_display.values, colLabels=cecl_display.columns, cellLoc='center', colLoc='center', loc='center')
table.auto_set_font_size(False)
table.set_fontsize(10)
table.scale(1.2, 2.1)

# Style header row
for (row, col), cell in table.get_celld().items():
    if row == 0:
        cell.set_text_props(weight='bold')
        cell.set_facecolor('#D9EAD3')

plt.tight_layout()

plt.savefig("../images/cecl_table.png", dpi=300, bbox_inches='tight')

plt.show()
```

```python
# ============================================================
# CECL MULTI-METRIC VISUALIZATION
# ============================================================

ax = cecl_table.set_index('grade')[['PD','LGD','EAD']].plot(kind='bar', figsize=(12,6), secondary_y='EAD', mark_right=False)

# Primary y-axis (PD and LGD)
ax.set_ylim(0, 1) 
ax.yaxis.set_major_formatter(PercentFormatter(1.0))
ax.set_ylabel('PD / LGD (%)', fontsize=12, fontweight='bold', fontname='Arial', labelpad=10)

# Secondary y-axis (EAD)
ax.right_ax.set_ylim(0, 18000)
ax.right_ax.yaxis.set_major_formatter(StrMethodFormatter('${x:,.0f}'))
ax.right_ax.set_ylabel('EAD ($)', fontsize=12, fontweight='bold', fontname='Arial', labelpad=10)

# x-axis (Grade)
ax.set_xlabel("Grade", fontsize=12, fontweight='bold', fontname='Arial', labelpad=10)
ax.tick_params(axis='x', labelrotation=0)

plt.title("CECL Risk Components by Grade", fontsize=18, fontweight='bold', fontname='Arial', pad=15)

# Formatting legend
handles1, labels1 = ax.get_legend_handles_labels()
handles2, labels2 = ax.right_ax.get_legend_handles_labels()
ax.legend(handles1 + handles2, labels1 + labels2, loc='upper center', bbox_to_anchor=(0.5, -0.15), ncol=3)

plt.tight_layout()

plt.savefig("../images/cecl_components.png", dpi=300, bbox_inches="tight")
plt.show()
```
