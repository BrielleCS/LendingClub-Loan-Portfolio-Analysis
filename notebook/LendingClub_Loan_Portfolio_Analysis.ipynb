# LendingClub Loan Portfolio Analysis: Realized Risk, Return, and Portfolio Concentration

## Overview
This analysis examines over 2.2M loans from the LendingClub accepted loans dataset issued between 2007–2018 to evaluate credit risk, loan performance, and portfolio concentration. To measure realized portfolio outcomes, the dataset was filtered to approximately 1.3M loans with finalized statuses — specifically Fully Paid and Charged Off loans — excluding loans still in repayment or other non-finalized statuses.

Using SQL for data extraction and aggregation, Python for data processing, and Tableau for visualization, the analysis focuses on key business questions related to borrower risk, portfolio returns, and concentration exposure.

The analysis identifies high-risk borrower segments, evaluates risk-adjusted returns across credit grades, and examines how loan exposure is distributed across borrower categories and geographic regions. The workflow demonstrates end-to-end analytical and visualization skills across the modern data analytics stack.

### Key Insights

**Underwriting Risk Trends** \
Default rates peaked at 26.6% in July 2016 origination cohort, indicating a period of elevated credit risk and weaker underwriting performance.

**Risk Distribution Across Borrower Segments** \
Defaults were broadly distributed across the portfolio, with 18 of 35 subgrades contributing to 80% of total defaults. Renters with Grade G loans exhibited the highest observed default rate at 54.9%.

**Risk vs. Return Trade-Off** \
A clear risk-return trade-off emerged across credit grades. Performance was strongest in higher grades, where both returns were positive and default risk remained relatively low. Grades A and B delivered the highest net returns (~5%) with single-digit to low-teens default rates. Grade C marked the inflection point, where default rates exceeded the portfolio benchmark (21.55%) while returns began to decline. Grade D continued this deterioration, with near break-even returns and materially elevated default risk. From Grade E onward, returns turned negative and default rates increased sharply (above 35%).

**Portfolio Concentration & Geographic Exposure** \
The portfolio exhibited moderate concentration risk (Gini Coefficient = 0.336), with significant exposure to debt consolidation loans (61.27% of volume, \\$11.9B) and borrowers in California ($2.9B exposure).

## Data Preparation

This section inspects the LendingClub dataset to understand its structure, size, and key variables before proceeding with this analysis.

Due to the large dataset size (2.2M+ rows), preprocessing was performed using a memory-efficient chunking approach in Python. The filtered dataset was then cleaned and prepared for all subsequent analyses.

### Key Steps:
- Loaded the LendingClub accepted loans dataset (151 columns) covering 2007–2018
- Applied filtering and column selection during chunked processing prior to exporting the reduced dataset
    - Selected 20 relevant columns for analyzing credit risk, returns, and portfolio concentration
    - Processed the dataset in 100k-row chunks to efficiently manage memory usage during preprocessing
    - Filtered the dataset to include only `Fully Paid` and `Charged Off` loans to ensure accurate measurement of default rates and realized returns, resulting in approximately 1.3M records  
- Handled missing values by imputing categorical variables (`emp_length`) with `Unknown` and retained missing date values (`last_pymnt_d`) as NaT, where appropriate for time-based analysis
- Converted date fields to datetime format and verified consistency across key numeric and categorical variables
- Performed validation checks on row counts and loan status distributions to ensure data consistency

```python
# -------------------------------
# Memory-Efficient CSV Filter
# -------------------------------

# Input and output paths
input_file = 'accepted_2007_to_2018Q4.csv'   # original large CSV
output_file = 'accepted_loans_filtered.csv'     # new filtered CSV

# Columns to keep
columns_to_keep = [
    'id', 'loan_amnt', 'term', 'int_rate', 'installment',
    'grade', 'sub_grade', 'emp_length', 'home_ownership', 
    'annual_inc', 'issue_d', 'loan_status', 'purpose', 
    'addr_state', 'fico_range_low', 'fico_range_high', 
    'total_pymnt', 'total_rec_prncp', 'total_rec_int', 'last_pymnt_d'
]

# Chunk size (100k rows at a time)
chunksize = 100_000

# Process chunks and write to new CSV
first_chunk = True  # to write header only once

for chunk in pd.read_csv(input_file, chunksize=chunksize, usecols=columns_to_keep):
    
    # filter for only 'Fully Paid' and 'Charged Off' loans
    chunk = chunk[chunk['loan_status'].isin(['Fully Paid', 'Charged Off'])]    
    
    # Append to new CSV
    chunk.to_csv(output_file, mode='w' if first_chunk else 'a', index=False, header=first_chunk)

    # after first chunk, header is no longer written
    first_chunk = False  

print(f"Filtered CSV saved as '{output_file}' successfully!")
```

```python
# Load filtered dataset
df = pd.read_csv("../data/accepted_loans_filtered.csv", low_memory=False)

# Handle missing categorical values
df['emp_length'] = df['emp_length'].fillna('Unknown')

# Convert date columns to datetime format; invalid values are coerced to NaT
df['issue_d'] = pd.to_datetime(df['issue_d'], errors='coerce')
df['last_pymnt_d'] = pd.to_datetime(df['last_pymnt_d'], errors='coerce')

# Export cleaned dataset
df.to_csv("accepted_loans_clean.csv", index=False, date_format='%Y-%m-%d')

#Create a small 1,000-row sample for GitHub
df.sample(n=1000, random_state=1).to_csv("../data/accepted_loans_sample.csv", index=False)

# Load cleaned dataset into PostgreSQL
df.to_sql('accepted_loans', engine, if_exists='replace', index=False)

# Validation checks
print("Loan status distribution:")
print(df["loan_status"].value_counts())

print("\nDataset size:")
print(len(df))

print("\nMissing values summary:")
print(df.isnull().sum().sort_values(ascending=False).head(10).to_string())
```

### Funded Volume & Default Rate by Grade Dataset

```python
query = """
WITH loan_data AS (
     SELECT grade, 
            SUM(loan_amnt) AS total_loan_exposure, 
            COUNT(*) FILTER (WHERE loan_status = 'Charged Off') AS charge_off_count,
            COUNT(*) AS total_count
     FROM accepted_loans
     GROUP BY grade
     )
SELECT grade,   
       total_loan_exposure,
       ROUND((charge_off_count * 100.0 / NULLIF(total_count,0)), 2) AS default_rate_pct
FROM loan_data
ORDER BY grade
"""

# SQL execution (kept for transparency; not run in final notebook)
# df_exposure_and_default = pd.read_sql(query, engine)
# df_exposure_and_default.to_csv("../data/exposure_and_default.csv", index=False)

# Load pre-generated dataset for reproducibility
df_exposure_and_default = pd.read_csv("../data/exposure_and_default.csv")
```

### Cohort Default Rate by Origination Month (2007–2018)
Monthly cohort-level default rate calculated using loan origination month (`issue_d`), measuring the proportion of loans that resulted in charge-off within each origination cohort.

```python
query = """
WITH loan_counts AS (
     SELECT DATE_TRUNC('year', issue_d) AS year,
            DATE_TRUNC('month', issue_d) AS month,
            COUNT(*) FILTER (WHERE loan_status = 'Charged Off') AS charge_off_count,
            COUNT(*) AS total_count
            FROM accepted_loans
            GROUP BY year, month
     )
SELECT year, 
       month, 
       ROUND((charge_off_count * 100.0 / NULLIF(total_count, 0)), 2) AS default_rate_pct
FROM loan_counts
ORDER BY year, month
"""

#df_default_rate_time = pd.read_sql(query, engine)
#df_default_rate_time.to_csv("../data/default_rate_pct_time.csv", index=False)
df_default_rate_time = pd.read_csv("../data/default_rate_pct_time.csv")
```

## Dashboard 2: Default Contribution & Risk Distribution

### Default Contribution Dataset (Pareto Construction)
Default contribution of realized losses across grades and subgrades used for Pareto analysis to assess whether a small number of grades and subgrades account for most of the total losses.

```python
query = """
WITH defaults AS (
     SELECT 
        grade, sub_grade,
        SUM(loan_amnt)::numeric AS default_amount
     FROM accepted_loans
     WHERE loan_status = 'Charged Off'
     GROUP BY grade, sub_grade
     ),
   cum_sum AS (
    SELECT
        grade, sub_grade,
        default_amount,
        default_amount * 1.0 / SUM(default_amount) OVER () * 100 AS default_contribution_pct,
        SUM(default_amount) OVER (ORDER BY default_amount DESC) / SUM(default_amount) OVER () * 100 AS cumulative_pct
    FROM defaults
    )
SELECT
    grade, sub_grade,
    ROUND(default_amount, 2) AS default_amount,
    ROUND(default_contribution_pct, 2) AS default_contribution_pct,
    ROUND(cumulative_pct, 2) AS cumulative_pct
FROM cum_sum
ORDER BY default_amount DESC
"""

# SQL execution (kept for transparency; not executed in final notebook)
# df_default_contribution = pd.read_sql(query, engine)
# df_default_contribution.to_csv("../data/default_contribution_pareto.csv", index=False)

df_default_contribution = pd.read_csv("../data/default_contribution_pareto.csv")
```

## Dashboard 3: Credit Risk & Return Analysis

### Credit Risk & Return Dataset
Grade-level aggregated dataset for return, risk, and interest rate metrics

```python
query = """ 
WITH loan_data AS (
     SELECT grade, 
            SUM(total_pymnt - loan_amnt) AS total_net_return,
            COUNT(*) FILTER (WHERE loan_status = 'Charged Off') AS charge_off_count,
            COUNT(*) AS total_count,
            (SUM(total_pymnt) - SUM(loan_amnt)) / SUM(loan_amnt) * 100 AS net_return_pct,
            AVG(int_rate) AS avg_int_rate
     FROM accepted_loans
     GROUP BY grade
     )
SELECT grade,
       total_net_return,
       net_return_pct,
       ROUND((charge_off_count * 100.0 / NULLIF(total_count, 0)), 2) AS default_rate_pct,
       avg_int_rate
FROM loan_data
ORDER BY grade
"""

# SQL execution (kept for transparency; not executed in final notebook)
# df_return_risk_int = pd.read_sql(query, engine)
# df_return_risk_int.to_csv("../data/return_risk_int.csv", index=False)

df_return_risk_int = pd.read_csv("../data/return_risk_int.csv")

# Weighted average interest rate across realized loans
df = pd.read_csv('../data/accepted_loans_clean.csv')
weighted_avg = np.average(df['int_rate'], weights = df['loan_amnt'])
print(f"Realized Average Interest Rate: {weighted_avg:.2f}%")
```

### Portfolio Concentration & Inequality (Lorenz Curve & Gini Coefficient)

Loan exposure distribution across borrowers is visualized using a Lorenz curve, highlighting the degree of concentration within the portfolio for Fully Paid and Charged Off loans. The Gini coefficient is computed to quantify inequality in loan exposure distribution, providing a single metric of concentration.

  The Gini Coefficient ranges from:\
    0 → Perfect equality (loan exposure evenly distributed across all borrowers)\
    1 → Perfect concentration (all loan exposure held by a single borrower)


```python
# Load cleaned dataset
df_lorenz = pd.read_csv("../data/accepted_loans_clean.csv")

# Prepare Lorenz curve data
loan_amnt_sorted = np.sort(df_lorenz['loan_amnt'])
cumulative_amnt = np.cumsum(loan_amnt_sorted)
cumulative_amnt = cumulative_amnt / cumulative_amnt[-1]
cumulative_id = np.arange(1, len(loan_amnt_sorted)+1) / len(loan_amnt_sorted)

# Plot Lorenze Curve
plt.figure(figsize=(12,6))
plt.plot(cumulative_id, cumulative_amnt, label='Lorenz Curve', color='blue')
plt.plot([0,1], [0,1], linestyle='--', color='gray', label='Perfect Equality')

# Shading Inequality Area
plt.fill_between(cumulative_id, cumulative_amnt, cumulative_id, color='orange', alpha=0.3, label='Inequality Area')

plt.title('Lorenz Curve - Loan Portfolio Concentration', fontsize=14, fontweight='bold', pad=15)
plt.xlabel('Cumulative Share of Borrowers', fontsize=12, fontweight='bold', labelpad=15)
plt.ylabel('Cumulative Share of Loan Amount', fontsize=12, fontweight='bold', labelpad=15)
plt.legend()
plt.grid(True, linestyle='--', alpha=0.5)

# Compute and plot Gini coefficient
gini = 1 - 2 * np.trapz(cumulative_amnt, cumulative_id)
plt.text(0.6, 0.2, f'Gini Coefficient: {gini:.3f}', fontsize=11, bbox=dict(facecolor='white', alpha=0.6))

#plt.savefig("../images/lorenz_curve.png", dpi=300, bbox_inches="tight")
#plt.show()

# Construct and save Lorenz curve dataset for later Tableau visualization
lorenz_df = pd.DataFrame({"cumulative_borrowers": cumulative_id, "cumulative_loan_amount": cumulative_amnt})
lorenz_df.to_csv("../data/lorenz_curve.csv", index=False)

# Save Gini coefficient
gini_df = pd.DataFrame({"gini_coefficient": [gini]})
gini_df.to_csv("../data/gini_coefficient.csv", index=False)
```
